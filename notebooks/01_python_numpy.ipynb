### Clening the data
The file metadata tells us:

- `SURF_FLAG == 0` means the surface-temperature value is considered good data.
- `NaN` means the temperature was not collected.

For this lesson, we will keep good surface-temperature observations and remove missing temperatures.

````python
sst = csv.loc[
    csv["SURF_FLAG"] == 0,
    ["YEAR", "MONTH", "DAY", "SURF_TEMP_C"]
].dropna(subset=["SURF_TEMP_C"]).copy()

sst.head()
````

### Creating a date column
Right now, year, month, and day are stored in separate columns. Pandas can combine them into a true **datetime** column.

Datetime columns make time-series filtering and plotting much easier.

````python
sst["DATE"] = pd.to_datetime(sst[["YEAR", "MONTH", "DAY"]])
sst = sst.sort_values("DATE")

sst.head()
````

````python
print("First date:", sst["DATE"].min())
print("Last date:", sst["DATE"].max())
````

## NumPy arrays: taking a column out of a DataFrame

A pandas **DataFrame** is useful because it keeps data in a labeled table. NumPy arrays are useful when we want to work directly with a sequence of numbers.

We can convert the cleaned temperature column into a NumPy array with `.to_numpy()

## `for` loops: repeating an operation

A **for loop** repeats the same block of code for every item in a sequence.

For example, we can print the first five SST observations one at a time:

````python
for t in temp[:5]:
    print(t)
````

Each time through the loop, `t` becomes the next temperature.

We can also loop over **indices** if we need to know both the position and the value.

````python
for i in np.arange(5):
    print("index =", i, "temperature =", temp[i])
````

### Combining a loop with an `if` statement

Now the computer can make a decision for every value.

````python
for t in temp[:10]:
    if t > 17:
        print(t, "°C  -> warmer than 17 °C")
    else:
        print(t, "°C  -> 17 °C or cooler")
````

### Using a counter

A common loop pattern is to count how many times something happens.

````python
warm_count = 0

for t in temp:
    if t > 20:
        warm_count += 1

print("Observations above 20 °C:", warm_count)
````

Notice that NumPy can do the same job in one line:

````python
np.sum(temp > 20)
````

Neither way is "wrong." The loop shows the logic step-by-step; NumPy is usually faster and more concise for large datasets.

## A more useful loop: compare month temperatures by year


````python
# isoalte aug obs
aug = sst[sst["MONTH"] == 8]

aug.head()
````

````python
# loop through a list of years and calculate the mean SST for each one
comparison_years = [1983, 1998, 2016, 2024, 2026]

for year in comparison_years:
    one_year = aug[aug["YEAR"] == year]
    mean_temp = one_year["SURF_TEMP_C"].mean()
    print(year, round(mean_temp, 2), "°C")
````

Sometimes we want to save each result instead of only printing it.

````python
# empty list
month_means_loop = []

for year in comparison_years:
    one_year = aug[aug["YEAR"] == year]
    mean_temp = one_year["SURF_TEMP_C"].mean()
    month_means_loop.append(mean_temp)

month_means_loop
````

## Grouping data with `groupby`

The loop above calculated one March mean at a time. Pandas can do the same calculation for **every year at once**.

````python
# make march df
march = sst[sst["MONTH"] == 3]

# groupby year
march_means = march.groupby("YEAR")["SURF_TEMP_C"].mean()

march_means.tail()
````

````python
plt.figure(figsize=(12, 4))
plt.plot(march_means.index, march_means.values)
plt.xlabel("Year")
plt.ylabel("Mean March SST (°C)")
plt.title("Mean March SST through time")
plt.show()
````

````python
# what are the warmest Marches on record?
march_means.sort_values(ascending=False).head(10)
````

````python
march_2026 = march_means.loc[2026]
march_rank = (march_means > march_2026).sum() + 1

print("March 2026 mean:", round(march_2026, 2), "°C")
print("Rank among March means:", march_rank, "out of", len(march_means))
````

## Anomalies: warmer or cooler than a reference period

An **anomaly** tells us how different an observation is from a reference value.

We will use **1991–2020** as the reference period and calculate a separate reference mean for January, February, and March.

````python
reference = sst[
    (sst["YEAR"] >= 1991) &
    (sst["YEAR"] <= 2020)
]

# 30 year mean for each month
reference_monthly = reference.groupby("MONTH")["SURF_TEMP_C"].mean()
reference_monthly.loc[[1, 2, 3]]
````

````python
means_2026 = sst_2026.groupby("MONTH")["SURF_TEMP_C"].mean()

comparison_2026 = pd.DataFrame({
    "MEAN_2026_C": means_2026,
    "REFERENCE_C": reference_monthly
}).loc[[1, 2, 3]]

comparison_2026["ANOMALY_C"] = (
    comparison_2026["MEAN_2026_C"] - comparison_2026["REFERENCE_C"]
)

comparison_2026
````

````python
plt.figure(figsize=(7, 4))
plt.bar(comparison_2026.index, comparison_2026["ANOMALY_C"])
plt.axhline(0, linewidth=1)
plt.xticks([1, 2, 3], ["January", "February", "March"])
plt.ylabel("SST anomaly (°C)")
plt.title("2026 SST anomalies relative to 1991–2020")
plt.show()
````

## Compare several El Niño winters with 2026

Now we can combine **groupby**, **filtering**, **plotting**, and another **for loop**.



````python
# calculate monthly means for every year and month
monthly_means = (
    sst.groupby(["YEAR", "MONTH"])["SURF_TEMP_C"]
       .mean()
       .reset_index()
)

monthly_means.head()
````

````python
# loop through comparison years and add a line to the same plot for each year
comparison_years = [1983, 1998, 2016, 2024, 2026]

plt.figure(figsize=(8, 5))

for year in comparison_years:
    one_year = monthly_means[
        (monthly_means["YEAR"] == year) &
        (monthly_means["MONTH"] <= 3)
    ]
    plt.plot(
        one_year["MONTH"],
        one_year["SURF_TEMP_C"],
        marker="o",
        label=str(year)
    )

plt.xticks([1, 2, 3], ["Jan", "Feb", "Mar"])
plt.xlabel("Month")
plt.ylabel("Mean SST (°C)")
plt.title("January–March SST")
plt.legend()
plt.show()
````
