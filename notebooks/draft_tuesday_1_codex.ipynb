## 1. How Codex talks to you

If you have seen GitHub Copilot, unlearn one thing first: **Codex does not finish your lines.** There is no grey ghost text and no `Tab` to accept — OpenAI has said inline autocomplete is not a direction they intend to take. Codex is an *agent*. You give it a task in the sidebar, it reads files, and it hands back an answer or a proposed edit for you to review.

| What you want | Where it happens |
|---|---|
| Ask a question about code | the composer at the bottom of the Codex sidebar |
| Give it a task that changes a file | the same composer — it replies with a diff |
| Decide what survives | that diff: accept, revise, or reject |

Three consequences of *"it reads files, not your screen"*:

1. **Save before you ask** (`Cmd+S` / `Ctrl+S`). Codex reads the file on disk. Unsaved cells do not exist as far as it is concerned, and neither does the output of a cell you have not re-saved.
2. **Point at things by name.** Highlighting a line and hoping does not reliably work from a notebook cell. Say where it is: *"In `notebooks/draft_tuesday_1_codex.ipynb`, explain the cell that calls `.isna().sum()`."*
3. **Pasting into the composer can misfire while a notebook is open** — VS Code may make a new notebook cell instead ([openai/codex#8599](https://github.com/openai/codex/issues/8599), still open). If that happens, undo with `Cmd+Z`, then either type the question by hand or click a plain `.py`/`.md` tab first so the notebook is not the active editor.

> **The one rule:** a line stays in your notebook only if you can say out loud what it does, without reading the code.

Reading your repository makes it sound more certain without making it better informed. That gap is the whole trap.
